````python
# imports

import os
from dotenv import load_dotenv
from openai import OpenAI
import subprocess
from IPython.display import Markdown, display
````

````python
load_dotenv(override=True)
groq_api_key = os.getenv('GROQ_API_KEY')
openrouter_api_key = os.getenv('OPENROUTER_API_KEY')

if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (and this is optional)")

if openrouter_api_key:
    print(f"OpenRouter API Key exists and begins {openrouter_api_key[:6]}")
else:
    print("OpenRouter API Key not set (and this is optional)")
````

````python
# Connect to client libraries

ollama_url = "http://localhost:11434/v1"
groq_url = "https://api.groq.com/openai/v1"
openrouter_url = "https://openrouter.ai/api/v1"

# Ollama ignores the API key, but the OpenAI client insists something be present
ollama = OpenAI(api_key="ollama", base_url=ollama_url)
groq = OpenAI(api_key=groq_api_key, base_url=groq_url)
openrouter = OpenAI(api_key=openrouter_api_key, base_url=openrouter_url)
````

````python
OLLAMA_MODEL_1 = "qwen2.5-coder"
OLLAMA_MODEL_2 = "deepseek-coder-v2"
OLLAMA_MODEL_3 = "gpt-oss:20b"
GROQ_MODEL = "openai/gpt-oss-120b"
OPENROUTER_MODEL = "qwen/qwen3-coder-30b-a3b-instruct"

# Want to try a different Ollama model? Run `ollama pull <model>` and swap the string above.
````

````python
compile_command = ["clang++", "-std=c++17", "-Ofast", "-mcpu=native", "-flto=thin", "-fvisibility=hidden", "-DNDEBUG", "main.cpp", "-o", "main"]
run_command = ["./main"]
````

## And now, on with the main task

````python
system_prompt = """
Your task is to convert Python code into high performance C++ code.
Respond only with C++ code. Do not provide any explanation other than occasional comments.
The C++ response needs to produce an identical output in the fastest possible time.
"""

def user_prompt_for(python):
    return f"""
Port this Python code to C++ with the fastest possible implementation that produces identical output in the least time.
The system information is:
{system_info}
Your response will be written to a file called main.cpp and then compiled and executed; the compilation command is:
{compile_command}
Respond only with C++ code.
Python code to port:

```python
{python}
```
"""
````

````python
def messages_for(python):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(python)}
    ]
````

````python
def write_output(cpp):
    with open("main.cpp", "w", encoding="utf-8") as f:
        f.write(cpp)
````

````python
def port(client, model, python):
    # Open source coding models are markedly more consistent at temperature=0 for this kind of task
    response = client.chat.completions.create(model=model, messages=messages_for(python), temperature=0)
    reply = response.choices[0].message.content
    reply = reply.replace('```cpp','').replace('```','')
    write_output(reply)
````

````python
pi = """
import time

def calculate(iterations, param1, param2):
    result = 1.0
    for i in range(1, iterations+1):
        j = i * param1 - param2
        result -= (1/j)
        j = i * param1 + param2
        result += (1/j)
    return result

start_time = time.time()
result = calculate(200_000_000, 4, 1) * 4
end_time = time.time()

print(f"Result: {result:.12f}")
print(f"Execution Time: {(end_time - start_time):.6f} seconds")
"""
````

````python
def run_python(code):
    globals = {"__builtins__": __builtins__}
    exec(code, globals)
````

````python
run_python(pi)
````

````python
port(ollama, OLLAMA_MODEL_1, pi)
````

# Compiling C++ and executing

This next cell contains the command to compile a C++ file based on the instructions above.

Again, it's not crucial to do this step if you don't wish to!

OR alternatively: you can run Python and C++ code online to test it out that way.
> Not an exact comparison but you can still get the idea of performance difference.
> For example here: https://www.programiz.com/cpp-programming/online-compiler/

````python
def compile_and_run():
    subprocess.run(compile_command, check=True, text=True, capture_output=True)
    print(subprocess.run(run_command, check=True, text=True, capture_output=True).stdout)
    print(subprocess.run(run_command, check=True, text=True, capture_output=True).stdout)
    print(subprocess.run(run_command, check=True, text=True, capture_output=True).stdout)
````

````python
compile_and_run()
````

### Notes on the open source swaps

- **No API key required at all** for the three Ollama models - everything runs on your own machine, at the cost of using your own CPU/GPU and being slower to respond than a hosted frontier model.
- **Groq and OpenRouter** are hosted, so they respond quickly like a frontier model would, but they're serving open weights models rather than closed ones - and both have generous free tiers.
- Smaller open source coding models occasionally produce C++ that doesn't compile on the first try, or that isn't actually faster than a naive translation - if a `compile_and_run()` call errors out, that's a real result worth recording too, not just a bug to fix.
